# Cleaning the dataset 'clubs'

```python
club_df = pd.read_csv('./data/clubs.csv')
club_df.head()
```

```python
club_df.info()
```

```python
club_df['average_age'].mean()
```

```python
# togliamo le colonne inutili
club_df = club_df.drop(columns=['total_market_value', 'coach_name', 'url'])

# i valori mancanti di average_age sono riempiti con -1
club_df['average_age'] = club_df['average_age'].fillna(-1)
```

```python
club_df[club_df['squad_size'] == 0].shape[0]
```

```python
# Calcola la percentuale di giocatori stranieri considerando il caso di squad_size zero
club_df['foreigners_percentage'] = np.where(
    club_df['squad_size'] > 0,
    ((club_df['foreigners_number'] * 100) / club_df['squad_size']).round(1),
    0
)

# Mostra il risultato aggiornato
print(club_df[['foreigners_number', 'squad_size', 'foreigners_percentage']])
```

```python
club_df.info()
```

```python
# convertiamo il campo net_transfert_record in un valore numerico
import re
def convert_currency_to_float(currency_str):
    if isinstance(currency_str, str):
        # Rimuovere il simbolo di euro e spazi aggiuntivi
        currency_str = currency_str.replace('€', '').strip()
        
        # Determinare il segno del numero
        sign = -1 if currency_str.startswith('-') else 1
        
        # Rimuovere segni di più e di meno per evitare conversioni errate
        currency_str = re.sub(r'[+-]', '', currency_str).strip()
        
        # Convertire milioni (m) e migliaia (k) in numeri float
        num = float(re.sub(r'[km]', '', currency_str))
        if 'm' in currency_str:
            num *= 1000000
        elif 'k' in currency_str:
            num *= 1000
            
        return sign * num
    return currency_str

club_df['net_transfer_record'] = club_df['net_transfer_record'].apply(convert_currency_to_float)
```

```python
club_df
```

```python
# Verifica dei record duplicati nel dataset
duplicates_count = club_df.duplicated().sum()

# Mostrare il numero di duplicati trovati
duplicates_count
```

```python
#Controllare che non ci siano stadi con dimensione negativa
negative_stadium_capacity = club_df[club_df['stadium_seats'] < 0]
negative_stadium_capacity.shape[0]
```

```python
# Controllare che la dimensione massima dello stadio sia un valore sensato
max_stadium_capacity = club_df['stadium_seats'].max()
max_stadium_capacity
```

```python
# Verificare che il numero dei giocatori stranieri sia minore del numero totale di giocatori nella squadra
invalid_foreigners = club_df[club_df['foreigners_number'] > club_df['squad_size']]
invalid_foreigners.shape[0]
```

```python
# Verificare che il numero di giocatori in nazionale sia minore del numero totale di giocatori nella squadra
invalid_nationals = club_df[club_df['national_team_players'] > club_df['squad_size']]
invalid_nationals.shape[0]
```

```python
# Verificare che la last season sia compresa tra il 2012 e il 2023
invalid_seasons = club_df[(club_df['last_season'] < 2012) | (club_df['last_season'] > 2023)]
invalid_seasons.shape[0]
```

```python
club_df.to_csv('./data_cleaned/clubs.csv', index=False)
```

# Cleaning the dataset 'players'

```python
players_df = pd.read_csv('./data/players.csv')
players_df.head()
```

```python
players_df.info()
```

```python
# Rimuoviamo le colonne specificate in quanto irrilevanti
columns_to_remove = ['image_url', 'url', 'first_name', 'last_name', 'highest_market_value_in_eur']
players_df = players_df.drop(columns=columns_to_remove)
```

```python
players_df.head()
```

```python
# Calcoliamo la percentuale di valori mancanti per ogni colonna
missing_data = players_df.isnull().mean() * 100

# Visualizziamo le percentuali di dati mancanti
missing_data.sort_values(ascending=False)
```

```python
# Impostiamo 'unknown' per i valori delle seguenti colonne:

players_df['agent_name'] = players_df['agent_name'].fillna('unknown')
players_df['country_of_birth'] = players_df['country_of_birth'].fillna('unknown')
players_df['city_of_birth'] = players_df['city_of_birth'].fillna('unknown')
players_df['country_of_citizenship'] = players_df['country_of_citizenship'].fillna('unknown')
players_df['sub_position'] = players_df['sub_position'].fillna('unknown')
```

```python
# inseriamo -1 nelle seguenti colonne per indicare che non hanno valori di mercato:
market_value_mean = players_df['market_value_in_eur'].mean()
players_df['market_value_in_eur'] = players_df['market_value_in_eur'].fillna(market_value_mean)
```

```python
# Calcoliamo le statistiche descrittive per alcune colonne numeriche rilevanti
stats = players_df['height_in_cm'].describe()

# Visualizziamo le statistiche per valutare la presenza di outlier
stats
```

```python
# Calcoliamo la mediana delle altezze, che useremo per sostituire il valore errato
height_median = players_df['height_in_cm'].median()

# Sostituiamo i valori errati di altezza (molto bassi, es. < 100 cm) con la mediana
players_df.loc[players_df['height_in_cm'] < 100, 'height_in_cm'] = height_median
```

```python
# Calcoliamo la mediana dell'altezza per ogni posizione
median_height_by_position = players_df.groupby('position')['height_in_cm'].median()

# Visualizziamo la mediana dell'altezza per ciascuna posizione
median_height_by_position
```

```python
# Imputiamo i valori mancanti di 'height_in_cm' basandoci sulla mediana della loro posizione
players_df['height_in_cm'] = players_df.groupby('position')['height_in_cm'].transform(lambda x: x.fillna(x.median()))

# Verifichiamo nuovamente la presenza di valori mancanti in 'height_in_cm' per assicurarci che siano stati tutti imputati
players_df['height_in_cm'].isnull().sum()
```

```python
# Visualizziamo i valori unici per alcune colonne categoriche per identificare incoerenze
unique_countries = players_df['country_of_birth'].unique()
unique_cities = players_df['city_of_birth'].unique()
unique_foot = players_df['foot'].unique()

unique_countries, unique_cities, unique_foot
```

```python
players_df['foot'] = players_df['foot'].fillna('unknown')
```

```python
# Dizionario per la mappatura dei nomi dei paesi
country_mapping = {
    "CSSR": "Czechoslovakia",
    "UdSSR": "Soviet Union",
    "East Germany (GDR)": "Germany",
    "United States": "USA",
    "United Kingdom": "UK"
}

# Applichiamo la mappatura al DataFrame
players_df['country_of_birth'] = players_df['country_of_birth'].replace(country_mapping)

# Verifichiamo nuovamente i valori unici per assicurarci che la mappatura sia stata applicata correttamente
sorted(players_df['country_of_birth'].unique())
```

```python
# Verifichiamo duplicati basati su 'name' e 'date_of_birth'
duplicate_players = players_df.duplicated(subset=['name', 'date_of_birth'], keep=False)
duplicated_data = players_df[duplicate_players].sort_values(by=['name', 'date_of_birth'])

# Visualizziamo i dati duplicati, se presenti
duplicated_data
```

```python
# rimuovo il doppione
# Troviamo l'indice della riga con player_id 461923 e rimuoviamo quella riga
index_to_drop = players_df[players_df['player_id'] == 461923].index
players_df_dropped = players_df.drop(index_to_drop)

# Verifichiamo se la riga è stata rimossa
players_df_dropped[players_df_dropped['player_id'] == 461923]
```

```python
# converto i valori nelle colonne delle date nel formato datetime
players_df['date_of_birth'] = pd.to_datetime(players_df['date_of_birth'], errors='coerce')
players_df['contract_expiration_date'] = pd.to_datetime(players_df['contract_expiration_date'], errors='coerce')
 
```

```python
missing_data = players_df.isnull().mean() * 100
missing_data
```

```python
players_df.to_csv('./data_cleaned/players.csv', index=False)
```

# Cleaning the dataset 'game_lineups' 

```python
# Caricamento del dataset
game_lineups_df = pd.read_csv('./data/game_lineups.csv')
game_lineups_df.head()
```

```python
game_lineups_df.info()
```

```python
# Conversione della colonna 'number' da stringa a numerico, gestendo valori non numerici e mancanti
game_lineups_df['number'] = pd.to_numeric(game_lineups_df['number'], errors='coerce')

# Verificare la presenza di valori mancanti nelle colonne
missing_values = game_lineups_df.isnull().sum()

# Ricerca di righe duplicate
duplicate_rows = game_lineups_df.duplicated().sum()

missing_values, duplicate_rows
```

```python
# Sostituire i valori NaN nella colonna 'number' con -1
game_lineups_df['number'] = game_lineups_df['number'].fillna(-1)

# Verificare la modifica
game_lineups_df['number'].isnull().sum()
```

```python
# Verificare la presenza di valori negativi nella colonna 'number' diversi da -1
negative_values_except_minus_one = game_lineups_df[game_lineups_df['number'] <= 0]
negative_values_except_minus_one = negative_values_except_minus_one[negative_values_except_minus_one['number'] != -1]

# Verificare la presenza di valori maggiori o uguali a 100 nella colonna 'number'
values_100_or_more = game_lineups_df[game_lineups_df['number'] >= 100]

negative_values_except_minus_one.shape[0], values_100_or_more.shape[0]
```

```python
# Verificare la presenza di valori diversi da 0 e 1 nella colonna 'team_captain'
invalid_team_captain_values = game_lineups_df[~game_lineups_df['team_captain'].isin([0, 1])]

invalid_team_captain_values.shape[0]
```

```python
# Verificare la presenza di valori diversi da "starting_lineup" e "substitutes" nella colonna 'type'
invalid_type_values = game_lineups_df[~game_lineups_df['type'].isin(["starting_lineup", "substitutes"])]

invalid_type_values.shape[0]
```
